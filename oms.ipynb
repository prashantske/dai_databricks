```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# Sample data
data = [
    (1, "Alice", 28),
    (2, "Bob", 34),
    (3, "Charlie", 29)
]

# Define schema
columns = ["id", "name", "age"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Show DataFrame
df.show()
```

```python
# Save the DataFrame as a managed Delta table in Unity Catalog
df.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("abc.def.tablename")  # Replace abc.def.tablename with your actual catalog and table
```
