```python
import json

# Define widget to receive result from Task 1
dbutils.widgets.text("result_json", "", "result_json")
result_json = dbutils.widgets.get("result_json")

# Print raw input for debugging
print(f"Raw result_json: {result_json}")



# Parse and use the JSON output
if result_json:
    try:
        result = json.loads(result_json)
        age = result["age"]
        department = result["department"]
        print(f"Age: {age}, Department: {department}")
    except json.JSONDecodeError as e:
        print("Error decoding JSON:", e)
else:
    print("No JSON input provided.")



```
