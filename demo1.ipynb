```python
import json


# Define widgets to receive input from the workflow
dbutils.widgets.text("name", "", "name")
dbutils.widgets.text("city", "", "city")
dbutils.widgets.text("age", "", "age")
dbutils.widgets.text("department", "", "department")

# Get widget values
name = dbutils.widgets.get("name")
city = dbutils.widgets.get("city")
age = dbutils.widgets.get("age")
department = dbutils.widgets.get("department")

print(name)
print(city)
print(age)
print(department)


```

```python

# Print some values
print(f"The city is {city} and the name is {name}")

# Prepare output data (only age and department)
output_data = {
    "age": age,
    "department": department
}

# Pass this data as output to next task
dbutils.notebook.exit(json.dumps(output_data))
```
